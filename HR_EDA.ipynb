```python
from sklearn.preprocessing import LabelEncoder
```

```python
# Load your HR dataset
df = pd.read_excel(r"C:\Users\Hp\OneDrive\Desktop\PROJECT 5\Excel_HR_Cleaned_Data.xlsx")
```

```python
# Step 1: Binary Encoding (LabelEncoder)
# Columns with 2 categories: Yes/No or Male/Female

binary_cols = ['Employee Attrition', 'Attrition Flag', 'Gender', 'Over Time']
le = LabelEncoder()
for col in binary_cols:
    df[col] = le.fit_transform(df[col])
```

```python
# Step 2: One-Hot Encoding for nominal categorical columns
# Columns with >2 categories and no natural order
# ----------------------------
nominal_cols = ['Department', 'Education Field', 'Job Role', 'Marital Status', 
                'Travel Frequency', 'Tenure Band', 'Income Band']
df = pd.get_dummies(df, columns=nominal_cols, drop_first=True)
```
